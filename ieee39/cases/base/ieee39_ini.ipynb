```python
import urllib.request

RAW = 'https://raw.githubusercontent.com/jmmauricio/benchmarks_public/main/ieee39/cases/base'
urllib.request.urlretrieve(f'{RAW}/ieee39.hjson', 'ieee39.hjson')
```

```python
import os
import time

from matplotlib import pyplot as plt
import numpy as np

os.environ['PYDAE_PARALLEL'] = '1'
os.environ['PYDAE_MAX_WORKERS'] = '10'

from pydae import ssa
from pydae.bps import BpsBuilder
from pydae.core.builder.casadi_builder import CasadiBuilder
from pydae.core.model.casadi_model import CasadiModel
from pydae.bps.utils.reporter import report_buses, report_gens, report_lines, report_all
from pydae.bps.utils.validator import validate_all, validate_buses, validate_gens



grid = BpsBuilder('ieee39.hjson', use_casadi=True)
grid.checker()
grid.uz_jacs = False
grid.construct('ieee39')
bld = CasadiBuilder(grid.sys_dict).build()
```

```python
model = CasadiModel(bld)
model.ini({}, newton_tol=1e-16)


report_buses(model, 'ieee39.hjson')
report_gens(model, 'ieee39.hjson')
validate_all(model, 'ieee39.hjson')

```

```python
model.A_eval()
ssa.damp_report(model)
```
